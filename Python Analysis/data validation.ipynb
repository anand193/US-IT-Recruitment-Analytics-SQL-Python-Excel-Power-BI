```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
applications = pd.read_csv("applications_clean.csv")
candidates = pd.read_csv("candidates_clean.csv")
companies  = pd.read_csv("companies_clean.csv")
interviews = pd.read_csv("interviews_clean.csv")
jobs      = pd.read_csv("jobs_clean.csv")
offers    = pd.read_csv("offers_clean.csv")
```

## Completeness

```python
print(candidates.isna().sum())
print(jobs.isna().sum())
print(applications.isna().sum())
print(interviews.isna().sum())
print(offers.isna().sum())
print(companies.isna().sum())
```

## Duplicate check

```python
print(candidates.duplicated().sum())
print(jobs.duplicated().sum())
print(applications.duplicated().sum())
print(interviews.duplicated().sum())
print(offers.duplicated().sum())
print(companies.duplicated().sum())
```

## checking primary key is unique

```python
print(candidates['candidate_id'].duplicated().sum())
print(jobs['job_id'].duplicated().sum())
print(applications['application_id'].duplicated().sum())
print(interviews['interview_id'].duplicated().sum())
print(offers['offer_id'].duplicated().sum())
print(companies['company_id'].duplicated().sum())
```

## Referential Integrity
**Referential integrity is a rule in a database that ensures relationships between tables remain valid and consistent**

```python
applications['candidate_id'].isin(
    candidates['candidate_id']
).value_counts()
```

```python
applications['job_id'].isin(
    jobs['job_id']
).value_counts()
```

```python
validation_summary = {
    'candidate_id_duplicates': candidates['candidate_id'].duplicated().sum(),
    'job_id_duplicates': jobs['job_id'].duplicated().sum(),
    'application_id_duplicates': applications['application_id'].duplicated().sum(),
    'interview_id_duplicates': interviews['interview_id'].duplicated().sum(),
    'offer_id_duplicates': offers['offer_id'].duplicated().sum(),
    'company_id_duplicates': companies['company_id'].duplicated().sum()
}

print(validation_summary)
```

```python
print(
    'Invalid candidate IDs:',
    (~applications['candidate_id'].isin(candidates['candidate_id'])).sum()
)

print(
    'Invalid job IDs:',
    (~applications['job_id'].isin(jobs['job_id'])).sum()
)

print(
    'Invalid interview application IDs:',
    (~interviews['application_id'].isin(applications['application_id'])).sum()
)

print(
    'Invalid offer application IDs:',
    (~offers['application_id'].isin(applications['application_id'])).sum()
)
```
